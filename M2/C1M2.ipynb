# Lecture: Essential Hardware and Software Components of Self-Driving Cars

## Introduction

This lecture focuses on the essential hardware and software components required to design and build a self-driving car. A modern autonomous vehicle is a complex system that combines **sensing, computing, control, and artificial intelligence** technologies. These systems work together to allow the vehicle to perceive its surroundings, make decisions, and safely navigate without direct human control.

A self-driving car can be divided into three major categories:

1. **Hardware Components** – Physical devices used for sensing, computation, movement, and power management.
2. **Programmable Hardware and Software** – Computing platforms and software frameworks that enable intelligence, decision-making, and control.
3. **Nonprogrammable Hardware Components** – Mechanical and electrical components that provide the physical structure and functionality of the vehicle.

---

# 1. Hardware Components of Self-Driving Cars

The hardware system of an autonomous vehicle acts as its body and sensory system. Sensors collect information about the environment, while onboard computers analyze the data and generate driving decisions.

## 1.1 Sensors

Sensors are critical components that allow a self-driving car to understand its surroundings. Different sensors provide different types of information, and multiple sensors are often combined to improve accuracy and reliability.

### Lidar (Light Detection and Ranging)

Lidar is a laser-based sensing technology used to create a detailed 3D map of the environment around the vehicle.

**Functions:**
- Measures the distance between the vehicle and surrounding objects.
- Generates accurate 3D point clouds of roads, buildings, vehicles, and pedestrians.
- Supports obstacle detection and autonomous navigation.
- Helps autonomous systems understand depth and spatial relationships.

Lidar is commonly used together with cameras and radar to improve environmental perception.

---

### Ultrasonic Sensors

Ultrasonic sensors use sound waves to detect nearby objects.

**Functions:**
- Detect objects at short distances.
- Assist with parking and low-speed maneuvering.
- Identify side obstacles and nearby vehicles.
- Provide additional safety information in tight spaces.

These sensors are inexpensive and effective for close-range detection.

---

### Cameras

Cameras provide visual information similar to human vision.

**Functions:**
- Capture images and video of the environment.
- Detect road signs, traffic lights, lane markings, vehicles, and pedestrians.
- Support computer vision algorithms.
- Enable object classification and scene understanding.

Advanced autonomous vehicles often use multiple cameras placed around the vehicle to provide a complete 360-degree view.

---

### Wheel Encoders

Wheel encoders measure wheel rotation to estimate movement.

**Functions:**
- Calculate vehicle speed.
- Measure distance traveled.
- Provide motion feedback to control systems.
- Support localization when combined with other sensors.

Wheel encoder data is especially useful when GPS signals are weak or unavailable.

---

### Radar

Radar uses radio waves to detect objects and measure their movement.

**Functions:**
- Detect vehicles and obstacles in different weather conditions.
- Measure object distance and speed.
- Operate effectively in rain, fog, and low visibility.
- Complement camera and lidar systems.

Radar is widely used because it performs reliably in environments where optical sensors may struggle.

---

### GPS (Global Positioning System)

GPS provides location information for navigation.

**Functions:**
- Determines the vehicle's global position.
- Supports route planning.
- Helps the vehicle follow predefined paths.
- Works together with mapping systems and localization algorithms.

Because GPS alone is not accurate enough for autonomous driving, it is usually combined with lidar, cameras, wheel encoders, and SLAM techniques.

# 1. Real-Time Operating Systems and Robot Operating System (ROS)

## 1.1 Real-Time Operating System (RTOS)

A Real-Time Operating System (RTOS) is an operating system designed to process tasks within strict time limits. It is essential for embedded devices used in self-driving cars because autonomous systems must respond quickly and reliably to changes in the environment.

Unlike traditional operating systems that prioritize general computing tasks, RTOS systems focus on predictable timing and fast responses.

### Functions of RTOS in Self-Driving Cars

- Provides real-time control of vehicle components.
- Ensures quick response from sensors and actuators.
- Manages low-level hardware operations.
- Supports safety-critical applications.
- Allows multiple tasks to run simultaneously with predictable timing.

Examples of RTOS applications include:
- Motor control.
- Brake and steering systems.
- Sensor communication.
- Safety monitoring systems.

Real-time performance is critical because delays in processing sensor information or controlling vehicle movement can affect safety.

---

# 1.2 Robot Operating System (ROS)

Robot Operating System (ROS) is an open-source framework widely used in robotics and autonomous vehicle development. Although its name includes "Operating System," ROS is actually a middleware framework that provides tools and libraries for building robotic applications.

ROS typically runs on Ubuntu Linux and provides communication between different software components of a robot or autonomous vehicle.

---

## ROS Communication Architecture

ROS uses a **publisher-subscriber communication model** to exchange information between different processes called nodes.

### Nodes

Nodes are individual software programs that perform specific tasks.

Examples:
- Camera processing node.
- Lidar processing node.
- Navigation node.
- Motor control node.

---

### Publisher-Subscriber Model

In this model:

- A **publisher node** sends information to a topic.
- A **subscriber node** receives information from that topic.

Example:

1. A camera node captures images.
2. The camera node publishes image data.
3. An object detection node subscribes to the image topic.
4. The AI model analyzes objects and sends decisions to the control system.

This modular structure makes autonomous vehicle software easier to develop, test, and maintain.

---

## Functions of ROS in Self-Driving Cars

- Provides communication between sensors and control systems.
- Supports autonomous navigation.
- Integrates hardware and software components.
- Provides tools for debugging and visualization.
- Enables robot automation and control.

# 2. Data Processing and Localization

Localization is one of the most important tasks in autonomous driving. The vehicle must know its exact position on a map before it can safely navigate.

---

# 2.1 Kinect Point Cloud Processing

The Kinect sensor generates 3D environmental data in the form of **point clouds**.

A point cloud is a collection of 3D points representing objects and surfaces around the vehicle.

### Processing Steps:

1. Kinect captures depth information.
2. The sensor generates a 3D point cloud.
3. The point cloud is processed using scanning algorithms.
4. The 3D information is converted into a 2D feature map.
5. The processed data is used by localization algorithms.

The conversion from 3D data to 2D representations reduces computational complexity while preserving important environmental information.

---

# 2.2 Adaptive Monte Carlo Localization (AMCL)

Adaptive Monte Carlo Localization (AMCL) is a probabilistic localization algorithm commonly used with ROS-based autonomous systems.

It determines the vehicle's position by combining:

- Odometry information from wheels.
- Depth information from sensors.
- A previously created map stored in ROS.

---

## How AMCL Works

1. The vehicle starts with an estimated position.
2. Multiple possible positions are generated as particles.
3. Sensor data is compared with the stored map.
4. Incorrect positions are removed.
5. The most likely position is selected.

This allows the vehicle to estimate its location even when sensor measurements are uncertain.

---

## Advantages of AMCL

- Provides accurate localization.
- Handles sensor noise.
- Works in complex environments.
- Updates position continuously while moving.

Accurate localization is essential because incorrect positioning can lead to navigation errors.

---

# 3. Path Planning and Control

After determining its current location, the self-driving car must calculate a safe path toward its destination.

---

# 3.1 Goal Position and Route Planning

The navigation system receives:

- Current vehicle location.
- Desired destination.
- Available map information.

The path planning algorithm then calculates the best route.

---

# 3.2 Static Shortest Path Algorithm

The static shortest path algorithm calculates the most efficient route between two points.

### Functions:

- Finds the shortest distance path.
- Reduces travel time.
- Provides a planned trajectory for the vehicle.

The generated path represents where the vehicle should move.

Other path planning algorithms can also be used depending on the environment, such as:

- A* algorithm.
- Dijkstra's algorithm.
- Rapidly-exploring Random Tree (RRT).
- Dynamic path planning methods.

---

# 3.3 Communication Between ROS and Arduino

Once a trajectory is generated, the command information is sent through ROS.

The communication process is:

1. The navigation algorithm calculates the desired path.
2. ROS receives the trajectory information.
3. ROS generates control commands.
4. Commands are sent to the Arduino board.
5. Arduino converts commands into motor signals.
6. Motors move the vehicle according to the planned path.

---

# 3.4 Motor Control

The Arduino controls vehicle movement by generating signals for the motors.

### Control Functions:

- Adjust motor speed.
- Control wheel rotation.
- Change vehicle direction.
- Maintain movement according to the planned trajectory.

This creates a connection between software decisions and physical vehicle actions.

---

# 4. Feedback and System Update

A self-driving car operates using a continuous feedback loop.

The vehicle constantly measures its movement and updates its understanding of the environment.

---

# 4.1 Feedback Loop Process

The complete cycle is:

1. Sensors collect environmental information.
2. Jetson Nano processes the sensor data.
3. Localization algorithms estimate vehicle position.
4. Path planning determines movement.
5. ROS sends commands to Arduino.
6. Arduino controls motors.
7. Wheel encoders measure actual movement.
8. Feedback is sent back to Jetson Nano.
9. The system updates its position and actions.

This cycle repeats continuously during operation.

---

# 4.2 Map Updating

The feedback information helps the vehicle maintain an accurate understanding of its environment.

Updates include:

- Vehicle position changes.
- Movement corrections.
- Updated sensor observations.
- Improved localization accuracy.

Continuous updates allow the vehicle to adapt to changing conditions.
