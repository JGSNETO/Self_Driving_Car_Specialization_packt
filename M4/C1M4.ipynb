# Computer Vision: From Visual Understanding to 3D Object Detection

## 1. What Is Computer Vision?

**Computer vision** is a field of artificial intelligence that enables machines to interpret and understand visual information from the physical world.

Humans naturally use their eyes, optic nerves, and visual cortex to understand their surroundings. Computer vision attempts to achieve a similar capability using:

* Cameras and other sensors
* Images and video
* Machine learning
* Deep learning
* Pattern recognition
* Neural networks

The goal is not simply to capture an image, but to extract **meaningful information** from it.

For example, a camera may capture an image of a road. A computer vision system can analyze that image and determine:

* Where the road is
* Where the lane markings are
* Whether a traffic light is present
* Which objects are cars
* Which objects are pedestrians
* Where those objects are located
* How those objects are moving

This ability to understand visual environments is fundamental to applications such as **self-driving cars, robotics, surveillance, agriculture, medical imaging, and augmented reality**.

---

# 2. How Computer Vision Understands Images

At a basic level, a digital image is represented as a collection of **pixels**.

A computer does not initially see an image as a human does. Instead, it receives numerical information representing properties such as color and intensity.

Deep learning models can learn patterns within these numerical representations.

One of the most important technologies used for image understanding is the **Convolutional Neural Network (CNN)**.

## Convolutional Neural Networks

CNNs are neural networks designed to process visual information efficiently.

Rather than manually programming every possible visual pattern, a CNN can learn useful features from training data.

For example, when learning to recognize a car, the network may gradually learn patterns corresponding to:

1. Edges
2. Corners
3. Shapes
4. Wheels
5. Windows
6. Larger object structures
7. The complete concept of a car

This is an example of **feature learning**.

The more representative training examples the model receives, the better it can generally learn to recognize objects in previously unseen images.

---

# 3. Pattern Recognition and Learning

A central idea behind computer vision is **pattern recognition**.

Suppose we want a system to distinguish between healthy and unhealthy crops.

We can provide the system with many labeled images:

```text
Image → Label

Healthy crop image   → Healthy
Unhealthy crop image → Unhealthy
Healthy crop image   → Healthy
...
```

The model analyzes these examples and learns patterns associated with each category.

During training, the model repeatedly compares its predictions with the correct answers and adjusts its internal parameters.

This iterative process allows the model to improve its predictions.

After training, we can provide a new image:

```text
New crop image → Computer Vision Model → Prediction
```

The model can then estimate whether the crop is healthy or unhealthy.

## Important Principle

> Computer vision models generally do not "understand" images in the same way humans do. They learn statistical patterns from visual data that allow them to make useful predictions.

# 4. What Can Computer Vision Do?

Computer vision systems can perform several different tasks.

### Classification

Classification answers:

> **What is in this image?**

For example:

```text
Image → "Dog"
```

The system assigns an image to one or more predefined categories.

---

### Object Identification

Identification determines what a particular object is.

For example, a system might identify:

```text
Object → Car
Object → Pedestrian
Object → Bicycle
```

---

### Object Detection

Detection answers two questions:

1. **What objects are present?**
2. **Where are they located?**

The output is commonly represented using **bounding boxes**.

For example:

```text
Image
┌─────────────────────────────┐
│                             │
│     ┌───────────┐           │
│     │    Car    │           │
│     └───────────┘           │
│                             │
│                  ┌──────┐   │
│                  │Person│   │
│                  └──────┘   │
└─────────────────────────────┘
```

---

### Object Tracking

Tracking extends detection across multiple frames.

Instead of simply detecting a car in one image, the system attempts to determine:

> **Where is the same car over time?**

For example:

```text
Frame 1 → Car detected at position A
Frame 2 → Same car detected at position B
Frame 3 → Same car detected at position C
```

This is particularly important for autonomous vehicles because objects are continuously moving.

---

# 5. Computer Vision vs. Human Vision

Humans and computer vision systems approach visual perception differently.

| Human Vision                                | Computer Vision                                              |
| ------------------------------------------- | ------------------------------------------------------------ |
| Uses eyes and optic nerves                  | Uses cameras and sensors                                     |
| Processes information through the brain     | Processes information through algorithms and neural networks |
| Learns from experience and observation      | Learns from training data                                    |
| Can generalize from relatively few examples | Often requires large datasets                                |
| Can become tired or distracted              | Can continuously process data                                |
| Naturally understands complex context       | Must learn patterns from data                                |

Computer vision has an important advantage: machines can process enormous amounts of visual data rapidly and consistently.

However, computer vision also has significant challenges.

A model that performs well on one dataset may perform poorly when the environment changes significantly.

For example:

```text
Training environment
        ↓
Sunny roads
        ↓
Model learns
        ↓
Deployment
        ↓
Rain + Snow + Night + Different Camera
        ↓
Performance may decrease
```

This is one reason why real-world computer vision requires diverse and representative datasets.

# 7. Computer Vision in Self-Driving Cars

Autonomous vehicles are one of the most demanding applications of computer vision.

A vehicle needs to continuously understand its environment.

For example, it may need to detect:

* Cars
* Trucks
* Motorcycles
* Pedestrians
* Cyclists
* Traffic lights
* Traffic signs
* Lane markings
* Road boundaries
* Obstacles

A simplified pipeline looks like this:

```text
Cameras / Sensors
       ↓
Visual Data
       ↓
Object Detection
       ↓
Object Tracking
       ↓
Environment Understanding
       ↓
Driving Decisions
```

Modern autonomous driving systems may use multiple cameras and additional sensors.

The goal is to create a reliable representation of the surrounding environment so the vehicle can make decisions in real time.

# 8. What Is Object Detection?

**Object detection** is the task of locating and identifying objects within an image or video.

Unlike image classification, which may assign a single label to an entire image, object detection provides information about individual objects.

For example:

```text
Input Image
     ↓
Object Detection Model
     ↓
┌──────────────────────────┐
│ Car       → Bounding Box │
│ Pedestrian → Bounding Box│
│ Bicycle   → Bounding Box │
└──────────────────────────┘
```

Object detection is essential in applications such as:

* Advanced Driver Assistance Systems (ADAS)
* Autonomous vehicles
* Video surveillance
* Robotics
* Image retrieval
* Industrial automation

---

# 9. 2D Object Detection

**2D object detection** operates primarily on images and identifies objects using 2D bounding boxes.

A bounding box can be represented approximately as:

```text
(x, y, width, height)
```

where:

* `x` = horizontal position
* `y` = vertical position
* `width` = bounding-box width
* `height` = bounding-box height

For example:

```text
Image
┌───────────────────────────────┐
│                               │
│       ┌──────────────┐        │
│       │     CAR      │        │
│       └──────────────┘        │
│                               │
└───────────────────────────────┘
```

The model can additionally provide a confidence score:

```text
Car → 96% confidence
```

---

# 10. Why 3D Object Detection?

2D detection tells us where an object appears in an image.

However, autonomous vehicles need more information about the **three-dimensional world**.

For example:

```text
2D detection:
"There's a car here."

3D detection:
"There's a car here,
approximately this far away,
with this size,
orientation, and 3D position."
```

A 3D detector may estimate a **3D bounding box** containing information such as:

* Position
* Width
* Height
* Length
* Orientation
* Depth

Conceptually:

```text
             3D Object
          ┌────────────┐
         /            /|
        /            / |
       └────────────┘  |
       |            |  |
       |            | /
       |            |/
       └────────────┘
```

This additional spatial information is extremely important for robotics and autonomous driving.

---

# 11. Sensors for 3D Object Detection

3D detection approaches can broadly be categorized according to the sensors they use.

Three major categories are:

1. **LiDAR-based**
2. **Camera-based**
3. **Multimodal / sensor fusion**

Each approach has different strengths and weaknesses.

---

# 12. LiDAR-Based 3D Object Detection

**LiDAR** produces a collection of 3D points representing the surrounding environment.

This collection is commonly called a **point cloud**.

A simplified point cloud can be visualized as:

```text
        •     •
    •       •       •
       •  •
  •          •
       •
            •      •
```

Each point can contain spatial information about where a surface exists.

LiDAR-based detection methods process these point clouds to identify objects and estimate their 3D bounding boxes.

---

## 12.1 Point-Based Methods

Point-based approaches operate directly on individual point-cloud points.

The model learns features from the points and uses them to predict objects.

Conceptually:

```text
Point Cloud
    ↓
Point Features
    ↓
Feature Learning
    ↓
3D Object Detection
```

---

## 12.2 Grid-Based Methods

Instead of processing every point independently, the point cloud can be organized into a structured representation such as a grid.

This can make processing more efficient and easier for certain neural network architectures.

---

## 12.3 Point-Voxel-Based Methods

A **voxel** is essentially a 3D equivalent of a pixel.

```text
Pixel → 2D space

Voxel → 3D space
```

Point clouds can therefore be converted into voxel representations.

A point-voxel method combines ideas from point-based and voxel-based processing.

---

## 12.4 Range-Based Detection

Another approach uses **range images** or related representations.

Instead of directly processing a raw 3D point cloud, the information can be represented according to distance from the sensor.

Features can also be transformed into representations such as:

* Bird's-eye view (BEV)
* Point-of-view representations

These representations can make the spatial structure of the environment easier for detection algorithms to process.

---

# 13. Camera-Based 3D Object Detection

Cameras are attractive for autonomous driving because they provide rich visual information and are generally less expensive than some specialized depth sensors.

Camera-based 3D detection can use different camera configurations.

### Monocular

Uses a single camera.

```text
Camera
   ↓
Single Image
   ↓
3D Object Estimation
```

The challenge is that a single image does not directly provide complete depth information.

The model therefore has to infer 3D information from visual cues.

---

### Stereo

Uses two cameras.

```text
Camera A ─┐
          ├─→ Stereo Information → Depth
Camera B ─┘
```

The difference between the views from the two cameras can be used to estimate depth through **stereo matching**.

---

### Multi-Camera

Multiple cameras can provide views from different directions around a vehicle.

This can improve environmental coverage and provide more information for 3D scene understanding.

---

# 14. Single-Stage vs. Two-Stage Detection

Deep learning object detectors are often broadly divided into **single-stage** and **two-stage** approaches.

### Two-Stage Detection

The system can conceptually be viewed as:

```text
Image
  ↓
Candidate Regions
  ↓
Object Classification + Localization
  ↓
Final Detections
```

The first stage proposes potential object regions.

The second stage analyzes those regions more carefully.

---

### Single-Stage Detection

Single-stage systems perform detection more directly:

```text
Image
  ↓
Detection Network
  ↓
Bounding Boxes + Classes
```

They are often attractive for applications where **real-time performance** is important.

---

# 15. YOLO and Real-Time Detection

**YOLO**, meaning **You Only Look Once**, is a well-known family of object detection models.

YOLO approaches detection as a direct prediction problem, making the architecture particularly useful for real-time applications.

YOLO models have been widely used for:

* Object detection
* Video analysis
* Real-time monitoring
* Autonomous systems
* Tracking pipelines

Different versions of YOLO have been developed over time, including **YOLOv7**.

For someone learning computer vision, understanding the general principles behind YOLO is useful because it demonstrates how modern deep learning models can perform fast object detection.

---

# 16. Multimodal 3D Object Detection

A single sensor may not provide all the information needed for reliable perception.

This leads to **multimodal detection**, where information from multiple sensors is combined.

For example:

```text
Camera ────────┐
               ├──→ Sensor Fusion → 3D Detection
LiDAR ─────────┘
```

The goal is to combine complementary information.

For example:

* Cameras provide rich appearance and texture information.
* LiDAR provides accurate spatial/depth information.

Together, these sources can potentially produce a stronger representation of the environment.

---

# 17. Early Fusion

In **early fusion**, information from different sensors is combined relatively early in the processing pipeline.

For example:

```text
Camera Data ──┐
              ├─→ Combined Representation → Detection
LiDAR Data ───┘
```

The combined data is then processed by the detection system.

---

# 18. Intermediate Fusion

In **intermediate fusion**, each sensor may first be processed independently.

```text
Camera → Feature Extraction ──┐
                              ├→ Feature Fusion → Detection
LiDAR  → Feature Extraction ──┘
```

Instead of directly combining raw sensor data, the system combines **learned features**.

This can provide greater flexibility because each sensor can have its own specialized feature-processing pipeline.

---

# 19. Temporal 3D Object Detection

Objects do not exist in isolation from one moment to another.

Vehicles move, pedestrians walk, and the camera itself moves.

Therefore, using information from multiple points in time can improve perception.

Instead of:

```text
Frame 1 → Detection
```

we can use:

```text
Frame 1
   ↓
Frame 2
   ↓
Frame 3
   ↓
Temporal Information
   ↓
Improved Detection / Tracking
```

Temporal methods can use sequences of LiDAR data or streaming sensor information.

This allows the system to reason about how objects and the environment change over time.

---

# 20. The Data Problem in Computer Vision

Modern deep learning models can require large amounts of labeled data.

For example, a dataset for autonomous driving may require humans to annotate:

```text
Car → 3D bounding box
Pedestrian → 3D bounding box
Cyclist → 3D bounding box
Traffic sign → Label
```

Creating these annotations can be expensive and time-consuming.

This creates an important research problem:

> **How can we build accurate computer vision systems without requiring enormous amounts of manually labeled data?**

Several learning strategies attempt to solve this problem.

---

# 21. Label-Efficient Learning

### Domain Adaptation

A model trained in one environment may not work equally well in another.

For example:

```text
Training Domain
     ↓
Clear weather + City A
     ↓
Model
     ↓
Target Domain
     ↓
Rain + City B
```

**Domain adaptation** attempts to reduce the performance gap between different domains.

---

### Weakly Supervised Learning

Instead of requiring detailed annotations for every object, weak supervision uses less precise or less expensive forms of labeling.

The goal is to learn useful models from incomplete or less detailed supervision.

---

### Semi-Supervised Learning

Semi-supervised learning combines:

```text
Small amount of labeled data
+
Large amount of unlabeled data
```

The model attempts to exploit both.

This is valuable because unlabeled images and sensor data can often be collected much more cheaply than manually annotated datasets.

---

### Self-Supervised Learning

Self-supervised learning attempts to generate useful training signals directly from the available data rather than relying entirely on human-provided labels.

This can allow models to learn representations from large quantities of unlabeled data.

---

# Advanced Computer Vision: Detection, Tracking, Segmentation, and Hydrant

This section extends the previous computer vision documentation with additional techniques for building efficient perception systems, particularly for autonomous vehicles.

---

# 1. Building Object Detectors

There are two major strategies for developing a deep-learning object detector.

## 1.1 Building a Detector from Scratch

A custom detector can be developed specifically for a particular problem.

This requires:

1. Designing the neural network architecture
2. Collecting a suitable dataset
3. Creating large numbers of labeled training examples
4. Training the model
5. Evaluating and optimizing its performance

This approach provides significant control over the architecture but can require substantial time, computational resources, and labeled data.

---

## 1.2 Using a Pretrained Detector

An alternative is to start with an existing pretrained object detection model.

Instead of learning everything from the beginning, the model can be adapted to a new task using **transfer learning**.

Conceptually:

```text
Pretrained Model
      ↓
Existing Learned Features
      ↓
Fine-Tuning
      ↓
New Detection Task
```

This can significantly reduce development time and the amount of training required compared with building an entirely new detector.

### Key idea

> Transfer learning allows knowledge learned from an existing dataset or task to be reused for a new computer vision problem.

---

# 2. Anchor Boxes in Single-Stage Detection

Single-stage detectors can use **anchor boxes** as predefined reference shapes for predicting objects.

Instead of predicting an arbitrary bounding box completely from scratch, the network can start from predefined box configurations and learn how to adjust them to match detected objects.

Conceptually:

```text
Image
  ↓
Anchor Boxes
  ↓
Box Adjustments
  ↓
Object Predictions
```

Anchor boxes can represent different expected object shapes and sizes.

This is particularly useful because objects in an image can have very different dimensions and aspect ratios.

---

# 3. Limitations of Single-Stage Detection

Although single-stage detectors are attractive because of their speed, they can face difficulties with **small objects**.

For example, a distant pedestrian may occupy only a small number of pixels:

```text
Large object
████████████████

Small object
██
```

When an object contains very little visual information, accurately detecting it becomes more challenging.

This represents an important trade-off in object detection:

```text
Higher speed  ↔  Detection accuracy
```

The appropriate balance depends on the application.

For autonomous vehicles, both speed and accuracy are extremely important because decisions must be made quickly while maintaining safety.

---

# 4. Traditional Machine Learning for Object Detection

Deep learning is not the only approach to object detection.

Before modern deep neural networks became dominant, computer vision systems commonly relied on **hand-designed features** combined with traditional machine learning algorithms.

Examples include:

* Aggregate Channel Features (ACF)
* Support Vector Machines (SVMs)
* Viola-Jones

---

## 4.1 Viola-Jones

The **Viola-Jones algorithm** became particularly well known for face detection.

It demonstrated that carefully designed visual features combined with machine learning could enable efficient object detection without deep neural networks.

---

## 4.2 Support Vector Machines

An **SVM** can be used as a classifier after extracting suitable features from an image.

A simplified traditional pipeline looks like:

```text
Image
  ↓
Hand-Crafted Feature Extraction
  ↓
Feature Vector
  ↓
SVM Classifier
  ↓
Object / Non-Object
```

The important distinction is that the feature representation is largely designed manually.

---

## 4.3 Traditional ML vs. Deep Learning

A major difference can be summarized as:

```text
Traditional ML:

Image
  ↓
Manually Designed Features
  ↓
Machine Learning Algorithm
  ↓
Prediction
```

versus:

```text
Deep Learning:

Image
  ↓
Neural Network
  ↓
Automatically Learned Features
  ↓
Prediction
```

Deep learning therefore reduces the need to manually determine which visual features are important.

---

# 5. Object Tracking as an Identity Problem

Object tracking introduces an additional requirement beyond detection:

> **The system must determine whether an object detected in one frame is the same object detected in another frame.**

Each detected object can therefore be assigned a unique identifier.

For example:

```text
Frame 1:
Car → ID 17

Frame 2:
Car → ID 17

Frame 3:
Car → ID 17
```

Even though the car changes position, the tracking system attempts to preserve its identity.

This is especially important when multiple objects of the same category are present.

```text
Frame:
Car A → ID 17
Car B → ID 42
Car C → ID 81
```

The system must avoid confusing these objects with one another.

---

# 6. Main Components of an Object Tracking Pipeline

A tracking system can be understood as three interconnected operations.

## Detection

Determine which objects are currently visible.

## Prediction

Estimate where an existing object is likely to appear in the next frame.

```text
Current Position
      ↓
Motion / Prediction
      ↓
Expected Next Position
```

## Data Association

Match newly detected objects with previously tracked objects.

```text
Previous Frame       Current Frame

Car ID 17  ───────→  Car ID 17
Car ID 42  ───────→  Car ID 42
```

Data association is what allows the system to maintain object identities over time.

---

# 7. Tracking Depending on Camera Motion

The tracking strategy can depend strongly on whether the camera is stationary or moving.

## Stationary Camera

When the camera remains fixed, the background tends to remain relatively stable.

This allows techniques such as **background subtraction** to identify regions that differ from the background.

```text
Static Background
       +
Moving Object
       ↓
Background Subtraction
       ↓
Candidate Moving Region
```

This can be useful for surveillance systems.

---

## Moving Camera

A moving camera causes the background itself to change between frames.

This makes simple background subtraction less suitable.

Approaches such as **sliding-window techniques** can instead be used to search for relevant visual patterns as the camera moves.

This distinction is important in autonomous vehicles because the cameras themselves are continuously moving through the environment.

---

# 8. Semantic Segmentation

Object detection identifies individual objects, but another important computer vision task is **semantic segmentation**.

Semantic segmentation assigns a class label to **every pixel** in an image.

For example:

```text
Pixel → Road
Pixel → Sidewalk
Pixel → Car
Pixel → Pedestrian
Pixel → Traffic Signal
Pixel → Background
```

The result is therefore a dense representation of the scene rather than a collection of bounding boxes.

Conceptually:

```text
Input Image
     ↓
Segmentation Network
     ↓
Pixel-by-Pixel Classification
     ↓
Semantic Scene Map
```

---

# 9. Why Semantic Segmentation Matters for Autonomous Driving

A self-driving vehicle needs to understand not only where individual objects are, but also the structure of the surrounding environment.

For example, segmentation can distinguish:

```text
┌─────────────────────────────┐
│           Sky               │
│                             │
│ Cars      Pedestrian        │
│                             │
│-----------------------------│
│          Road               │
│       Drivable Area         │
└─────────────────────────────┘
```

This allows a vehicle to identify areas that are suitable for driving and distinguish them from sidewalks, obstacles, and other scene regions.

Semantic segmentation is also highly transferable to other domains, including **medical image analysis**, where individual pixels may represent different anatomical structures or tissue types.

---

# 10. Mathematical View of Semantic Segmentation

Semantic segmentation can be viewed as a function that produces a score for every possible class at every pixel.

For a pixel \(p\), the model can produce:

```text
Road       → 0.82
Sidewalk   → 0.10
Car        → 0.05
Pedestrian → 0.03
```

The predicted class is the class with the highest score:

```text
Prediction = argmax(class scores)
```

Therefore, segmentation produces a classification decision independently for each pixel while considering the surrounding image context.

---

# 11. Challenges in Semantic Segmentation

Semantic segmentation is difficult because visual appearance can vary substantially.

Important challenges include:

### Occlusion

One object can partially hide another.

```text
Person A
   █████
    ████
Person B
   █████
```

The model must infer the correct segmentation even when only part of an object is visible.

### Scale Variation

Objects can appear at very different sizes.

A nearby car may occupy a large portion of an image, while a distant car may occupy only a few pixels.

### Illumination Changes

The same object can look very different depending on:

* Bright sunlight
* Shadows
* Night conditions
* Reflections
* Weather

### Ambiguous Boundaries

Some objects have boundaries that are difficult to distinguish, particularly when:

* Objects have similar colors
* Objects overlap
* The image resolution is limited
* Objects are far away

These challenges make pixel-level prediction significantly more demanding than simple image classification.

---

# 12. Evaluating Semantic Segmentation with IoU

A common metric for semantic segmentation is **Intersection over Union (IoU)**.

IoU measures the overlap between the predicted region and the ground-truth region.

For a particular class:

```text
IoU = TP / (TP + FP + FN)
```

where:

* **TP** = True Positives
* **FP** = False Positives
* **FN** = False Negatives

Conceptually:

```text
             Ground Truth
          ┌───────────────┐
          │      █████    │
          │   █████████   │
          └───────────────┘
                 ∩
             Prediction
```

A larger overlap produces a higher IoU.

An IoU close to 1 indicates strong agreement between prediction and ground truth.

---

# 13. Important IoU Evaluation Practices

There are several important details when calculating segmentation performance.

## Aggregate Across Images

IoU should be calculated over all relevant images collectively rather than simply calculating an IoU for every image and averaging those values.

This produces a more meaningful aggregate measurement.

## Prefer Per-Class IoU

Performance should also be examined separately for each class.

For example:

```text
Road       → 92% IoU
Car        → 85% IoU
Pedestrian → 71% IoU
Sidewalk   → 88% IoU
```

This is more informative than relying only on a single global score because some classes may be much harder to segment than others.

# 16. Why U-Net Improves on Sliding Windows

A traditional sliding-window approach processes many overlapping regions independently.

For example:

```text
┌──────┐
│Patch │
└──────┘

   ┌──────┐
   │Patch │
   └──────┘

      ┌──────┐
      │Patch │
      └──────┘
```

Because neighboring patches overlap, the same pixels may be processed repeatedly.

This creates **redundant computation** and can make training slow.

U-Net instead processes the image using the network architecture to capture information from the entire image context.

Conceptually:

```text
Entire Image
     ↓
   U-Net
     ↓
Pixel-Level Segmentation
```

This provides better computational efficiency and allows the model to use broader contextual information when making pixel-level predictions.

# 17. Hydrant: Dynamic Neural Network Architecture

**Hydrant** is a dynamic neural network architecture designed for efficient semantic segmentation and regression tasks.

Its key idea is to use multiple specialized **branches**, also called **heads**.

Instead of requiring one large network to perform every task in exactly the same way, different branches can be trained for different tasks or subsets of data.

Conceptually:

```text
                    Input
                      │
                Shared Network
                      │
             ┌────────┼────────┐
             ↓        ↓        ↓
          Branch 1  Branch 2  Branch 3
             │        │        │
          Task A    Task B    Task C
```

This architecture provides a way to make neural networks more computationally efficient during inference.

---

# 18. The Hydrant Gate Mechanism

A key component of Hydrant is its **gate**.

The gate determines which branches should be activated for a particular input.

```text
                  Input
                    ↓
                  Gate
             ┌──────┼──────┐
             ↓      ↓      ↓
          Branch A Branch B Branch C
             ✓       ✗       ✓
              \      |      /
               \     |     /
                 Combined
                    ↓
                Prediction
```

Rather than activating every branch for every input, the architecture can selectively activate appropriate branches.

This creates a form of **dynamic computation**.

### Core idea

> The network does not necessarily perform the same amount of computation for every input; the gate helps determine which specialized components are needed.

# 21. Bird's-Eye-View Representation

A **bird's-eye view (BEV)** represents the environment from above.

Instead of reasoning only from the perspective of an individual camera, the system can transform information into a spatial representation around the vehicle.

Conceptually:

```text
             Front
               ↑
       ┌───────────────┐
       │     Car       │
       │               │
       │   ┌───────┐   │
       │   │Vehicle│   │
       │   └───────┘   │
       │               │
       └───────────────┘
               ↓
             Rear
```

BEV representations are useful because they provide a more direct spatial description of the environment around an autonomous vehicle.

# 22. Hydrant and Data Augmentation

Another important concept associated with Hydrant is a data augmentation strategy for **regression neural networks**.

Deep learning models often require large quantities of labeled data.

In domains such as medical image analysis, obtaining large labeled datasets can be particularly difficult because expert annotation may be expensive.

Hydrant uses a method that can **recombine existing labeled samples** to create additional training examples.

Conceptually:

```text
Labeled Sample A ──┐
                   ├─→ Recombination → New Training Sample
Labeled Sample B ──┘
```

The resulting augmented data can help regularize neural network regressors.

---

# 23. Why Data Augmentation Matters

Training a neural network on a limited dataset can cause the model to memorize the available examples instead of learning general patterns.

Data augmentation provides additional variation during training.

```text
Small Labeled Dataset
        ↓
Data Augmentation
        ↓
More Diverse Training Examples
        ↓
Better Generalization
```

The Hydrant approach demonstrates that augmentation does not necessarily have to consist only of simple transformations such as flipping, cropping, or rotation.

Existing labeled samples can also be recombined in meaningful ways.

The lecture describes its application to tasks such as estimating features from brain images.

# 24. Connecting the New Concepts

The techniques introduced in this section address different levels of the computer vision problem:

```text
Object Detection
      ↓
Where are the objects?

Object Tracking
      ↓
Which object is which over time?

Semantic Segmentation
      ↓
What does every pixel represent?

3D / BEV Understanding
      ↓
Where are things in the surrounding space?

Dynamic Architectures
      ↓
How can perception be performed efficiently?

Data Augmentation
      ↓
How can models learn effectively with limited labeled data?
```

Together, these techniques illustrate that modern computer vision is not just about recognizing objects.

A complete perception system must also understand **pixel-level scene structure, object identity over time, 3D spatial relationships, computational efficiency, and how to learn effectively from limited data**.
